```python
%%writefile 7_button_demp.py
import streamlit as st
import pandas as pd

if st.button("Click here to Generate sample data"):
    df=pd.DataFrame({
        'Enrollment number ':[1,2,3,4],
        'Marks ':[78,85,92,69]
    })
    st.write("Generated Date")
    st.dataframe(df)
    csv=df.to_csv(index=False).encode('utf-8')
    st.download_button(label='Download as csv',data=csv,file_name='sample.csv',mime='text/csv')
```

```python
%%writefile 8_display_data.py
import streamlit as st
import pandas as pd
st.title("Displaying a Data in streamlit")

data={'Student ':['A','B','C','D'],
     'marks ':[85,92,69,87],
     'passed ':[True,True,True,True]}
df=pd.DataFrame(data)
st.subheader("st.dataframe() (interactive)")
st.dataframe(df)

st.subheader("st.table() (static)")
st.table(df)

st.subheader("st.json()(Structured json)")
st.json(df)
```

```python
%%writefile 10_status_demo.py
import streamlit as st
import time

st.title('Status Element Demo')

st.success('This is Success Message')
st.warning('This is Warning Message')
st.error('This is Error Message')
st.info('Usefull information can go here')

st.write('----')
st.subheader('Progress & spinner Example')

if st.button('Start long task'):
    progress=st.progress(0)
    with st.spinner('Progressing.....'):
        for i in  range(100):
            time.sleep(0.1)
            progress.progress(i+1)
    st.success("Task compelete")
```

```python
%%writefile 11_matplotlib.py
import streamlit as st
import matplotlib.pyplot as plt
import numpy as np

st.title('MATplotlib + Streamlit Demo')

x=np.arange(1,11)
y=np.random.randint(50,100,size=10)

#---------------------------------------
#Line Chart
#---------------------------------------

st.subheader('Line chart')
plt.figure(figsize=(4,2))
plt.plot(x,y,marker='o')
plt.xlabel('Student Index')
plt.ylabel('Marks')
plt.title('Marks of 10 students')
st.pyplot(plt)
plt.clf()

#---------------------------------------
#Bar Chart
#---------------------------------------

st.subheader('Bar chart')
plt.figure(figsize=(4,2))
plt.bar(x,y)
plt.xlabel('Student Index')
plt.ylabel('Marks')
plt.title('Marks of 10 students')
st.pyplot(plt)
```

```python
%%writefile 12_matplotlib.py
import streamlit as st
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

st.title('Matplotlib Streamlit Demo')
marks=np.random.randint(50,100,size=10)
attendance=np.random.randint(50,100,size=10)

chart_data=pd.DataFrame({
    'Marks':marks,
    'Attendance':attendance
})

st.subheader('Line chart')
st.line_chart(chart_data)

st.subheader('Area chart')
st.area_chart(chart_data)

st.subheader('Bar chart - Histogram Approximation')
counts,bins=np.histogram(marks,bins=5)
st.bar_chart(counts)
```
