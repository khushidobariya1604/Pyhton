```python
import matplotlib.pyplot as plt
import numpy as np
time=np.arange(12)
income=np.array([5,9,6,6,10,7,6,4,4,5,6,4])
expance=np.array([6,6,8,3,6,9,7,8,6,6,4,8])
plt.plot(time,income,color='green')
plt.plot(time,expance,color='red')
plt.fill_between(time,income,expance, where=(income>expance),color="green",alpha=0.4)
plt.fill_between(time,income,expance, where=(income<expance),color="red",alpha=0.4)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
time=np.arange(12)
income=np.array([5,9,6,6,10,7,6,4,4,5,6,4])
expance=np.array([6,6,8,3,6,9,7,8,6,6,4,8])
plt.plot(time,income,color='green')
plt.plot(time,expance,color='red')
plt.fill_between(time,income,expance, where=(income>expance),color="green",alpha=0.4,label='Positive',interpolate=True)
plt.fill_between(time,income,expance, where=(income<expance),color="red",alpha=0.4,label='Negative',interpolate=True)
plt.xlabel=('time')
plt.ylabel=('income/expance comparision')
plt.title=('Area fill between two lines')
plt.legend()
plt.show()
```
