```python
import numpy as np
import sys

s=range(1000)

print("Size of each element of list in bytes : ",sys.getsizeof(s))
print("Size of whole list in bytes : ",sys.getsizeof(s)*len(s))

d=np.arange(1000)

print("Size of each element of numpy array in bytes : ",d.itemsize)
print("Size of whole numpy array in bytes : ",d.size*d.itemsize)
```

**> Reshape :**

```python
import numpy as np
y=np.array([1,2,3,4,5,6])
newarr=y.reshape(6,1)
print(newarr)
```

```python
newarr=y.reshape(1,6)
print(newarr)
```

```python
newarr=y.reshape(2,3)
print(newarr)
```

```python
newarr=y.reshape(3,2).base
print(newarr)
```

```python
import numpy as np
d=np.array([1,2,3,4,5,6])
newarr=d.reshape(2,-1)
print(newarr)
```

```python
newarr=d.reshape(-1,3)
print(newarr)
```
