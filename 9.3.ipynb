**> Concatenate() :**

```python
import numpy as np
arr1=np.array([1,2,3])
arr2=np.array([4,5,6])
arr3=np.array([7,8,9])
arr=np.concatenate ((arr1,arr2,arr3))
print(arr)
```

```python
import numpy as np
arr1=np.array([[1,2,3],[4,5,6]])
arr2=np.array([[7,8,9],[10,11,12]])
arr=np.concatenate ((arr1,arr2))
print(arr)
```

```python
import numpy as np
arr1=np.array([[1,2,3],[4,5,6]])
arr2=np.array([[7,8,9],[10,11,12]])
arr=np.concatenate ((arr1,arr2),axis=0)
print(arr)
```

```python
import numpy as np
arr1=np.array([[1,2,3],[4,5,6]])
arr2=np.array([[7,8,9],[10,11,12]])
arr=np.concatenate ((arr1,arr2),axis=1)
print(arr)
```

```python
from numpy import random
x=random.randint(1,100)
def guess():
    attempt=7
    for i in range(1,8):
        n=int(input("Enter the number between 1 to 100 : "))
        print("Attempt number = ",i)
        if n==x:
            print("Correct number")
            print("You won")
            break
        else:
            attempt=attempt-1
            print("Remain attempt= ",attempt)
            print()
            if attempt==0:
                print("You lose the game")
                print()
                print('Correct number is : ',x)
guess()
```
