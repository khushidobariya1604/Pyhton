**1. Polymorphism:**
- allows programmers to write methods that can process a variety of different types of functionalities with the same name.
  
  1. overloading : implies using numerous methods with the same name but accepting a different no. of arguments.(compiletime polymorphism)
  2. overriding : child class method overrides the parent class method of the name,parameters and return type. it is used                     to overwrite a parent class method in derived class.(runtime polymorphism)

```python
def sum_numbers(*args):
    result=0
    for i in args:
        result+=i
    print("Sum = ",result)
print("Similar to method overloading concept")
print("single argument = ",end=" ")
sum_numbers(10)
print("two argument = ",end=" ")
sum_numbers(30,2)
print("Multiple argument = ",end=" ")
sum_numbers(1,2,3,4,5)
```

**Operator overloading:**
- giving extented meaning beyond their predefined operational meaning

```python
class A:
    def __init__(self,a):
        self.a=a
    def __add__(self,other):
        return self.a+other.a
obj1=A(1)
obj2=A(2)
obj3=A('hello')
obj4=A('Hii')
```

```python
print(obj1+obj2)
print(obj3+obj4)
```

**2. Inheritance:**
- We can use existing class to create new class rather than creating from scratch.
- inheritance allows the program to reuse the code
- with inheritance. child class gains access to all the data members,functions and properties defined in parent class
- A child class may also offer it's particular implementations of parent class functions

**1. Single Inheritance (simple Inheritance):**

```python
class parent:
    def fun1(self):
        print("Hello Parent")
class child(parent):
    def fun2(self):
        print("Hello child")
obj=child()
obj.fun1()
obj.fun2()
```

**2. Multiple Inheritance:**

```python
class parent1:
    def func1(self):
        print("Hello Parent1")
class parent2:
    def func2(self):
        print("Hello Parent2")
class parent3:
    def func2(self):
        print("Hello Parent3")
class child(parent1,parent3,parent2):
    def func3(self):
        print("Hello child")
obj=child()
obj.func1()
obj.func2()
obj.func3()
print(child.__mro__)
```

```python
class Grandparent:
    def fun1(self):
        print("Hello Grandparent")
class parent(Grandparent):
    def fun2(self):
        print("Hello parent")
class child(parent):
    def fun3(self):
        print("Hello child")
obj=child()
obj.fun1()
obj.fun2()
obj.fun3()
```

**3. Hierarchical Inheritance :** 

```python
class parent:
    def func1(self):
        print("Hello Parent")
class child1(parent):
    def func2(self):
        print("Hello child1")
class child2(parent):
    def func3(self):
        print("Hello child3") 
obj1=child1()
obj2=child2()
obj1.func1()
obj1.func2()
print("----------------")
obj2.func1()
obj2.func3()
```

**4. Hybrid Inheritance:**

```python
class parent1:
    def func1(self):
        print("Parent 1")
class parent2:
    def func2(self):
        print("Parent 2")
class child1(parent1):
    def func3(self):
        print("child1")
class child2(child1,parent2):
    def func4(self):
        print("child1")
o1=child1()
o2=child2()

o1.func1()
o1.func3()
print("--------------")
o2.func1()
o2.func2()
o2.func3()
o2.func4()
```

1. super()

```python
class parent:
    def __init__(self):
        self.atr1=50
        self.atr2=60
class child(parent):
    def __init__(self):
        super().__init__()  # or  parent.__init__(self)
        self.atr3=49
obj1=child()
print(obj1.atr1)
```

```python
class parent:
    def display(self):
        print("Hello parent")
class child(parent):
    def display(self):
        super().display()
        print("Hello child")
obj=child()
obj.display()
```

```python
class parent:
    def func1(self):
        print("Hello parent")
class child(parent):
    def func2(self):
        print("Hello child")
print(issubclass(child,parent))
print(issubclass(parent,child))
```

**> Method resulution order (MRO) :**

- denotes the way a program. language resolves a method or attribute
- MRO defines the order in which the base classes are searched when executing a method
- first, the method or attribute is searched within a class and then it follows the order we specified while inheriting

```python
class A:
    def rk(self):
        print("class A")
class B(A):
    def rk(self):
        print("class B")
class C(A):
    def rk(self):
        print("Class C")
class D(B,C):
    pass
obj=D()
obj.rk()
print(D.mro())
```

**> Abstract Class:**

```python
from abc import ABC,abstractmethod
class shape(ABC):
    def __init__(self,shape_name):
        self.shape_name=shape_name
    @abstractmethod
    def draw(self):
        pass
class circle(shape):
    def __init__(self):
        super().__init__("circle")
    def draw(self):
        print("Drawing :",self.shape_name)
obj=circle()
obj.draw()
```

```python
from abc import ABC,abstractmethod
class shape(ABC):
    def __init__(self,shape_name):
        self.shape_name=shape_name
    @abstractmethod
    def draw(self):
        pass
class circle(shape):
    def __init__(self):
        super().__init__("circle")
    def draw_circle(self):
        print("Drawing :",self.shape_name)
    def draw(self):
        pass
obj=circle()
obj.draw_circle()
```

- book --> name,n,author,publisher,ISBN,year
- coursebook -->name,n,author,publisher,ISBN,year, course,display()

```python
class Book:
    def __init__(self,name,n,author,publisher,ISBN,year):
        self.name=name
        self.n=n
        self.author=author
        self.publisher=publisher
        self.ISBN=ISBN
        self.year=year
    def display(self):
        print("Book name = ",self.name)
        print("Number of authors = ",self.n)
        print("Author name : ",self.author)
        print("Book Publisher : ",self.publisher)
        print("ISNR no. : ",self.ISBN)
        print("Year : ",self.year)
class Coursebook(Book):
    def __init__(self,name,n,author,publisher,ISBN,year,course):
        super().__init__(name,n,author,publisher,ISBN,year)
        self.course=course
    def display(self):
        super().display()
        print("Course name : ",self.course)
obj=Coursebook('Herry Potter',1,['john'],'alex',96581,2020,'CSE')        
obj.display()
```
